1. Write a Python function compute_gradient(f, x) that takes a simple function f (like f(x) = x**2) and a value x, and returns the numerical gradient at x using the central difference method.<br><br><em><strong>Hint:</strong> Use (f(x+h) - f(x-h)) / (2*h) with a small h (e.g., 1e-5).</em>

```python
import math
```

```python
def compute_gradient(f, x):
    h = 1e-5
    gradient = (f(x + h) - f(x - h)) / (2 * h)
    return gradient

def f(x):
    return x ** 2
```

```python
x=3
```

```python
gradient = compute_gradient(f, x)
```

```python
print("Value of x:", x)
print("Numerical Gradient:", gradient)
```

2. Given the function f(x, y) = x*2 + y*2, calculate the partial derivatives with respect to x and y at the point (x=3, y=4) using Python.

```python
def f(x, y):
    return x**2 + y**2

def partial_derivatives(x, y):
    h = 1e-5

    df_dx = (f(x + h, y) - f(x - h, y)) / (2 * h)
    df_dy = (f(x, y + h) - f(x, y - h)) / (2 * h)

    return df_dx, df_dy
```

```python
x = 3
y = 4
```

```python
dx, dy = partial_derivatives(x, y)
```

```python
print("Point: (x =", x, ", y =", y, ")")
print("Partial derivative with respect to x:", dx)
print("Partial derivative with respect to y:", dy)
```

4. Create a Python script that demonstrates how different learning rates (0.01, 0.1, 0.5) affect the convergence of gradient descent for f(x) = (x-3)**2. Plot the value of x over 20 iterations for each learning rate using matplotlib.<br><br><em><strong>Hint:</strong> Use a loop to update x for each learning rate and plot all three on the same graph.</em>

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return (x - 3) ** 2

def gradient(x):
    return 2 * (x - 3)
```

```python
learning_rates = [0.01, 0.1, 0.5]
x0 = 0
iterations = 20
```

```python
plt.figure(figsize=(8, 6))

for lr in learning_rates:
    x = x0
    x_values = [x]

    for i in range(iterations):
        grad = gradient(x)
        x = x - lr * grad
        x_values.append(x)

    plt.plot(range(iterations + 1), x_values, label=f"Learning Rate = {lr}")

plt.title("Gradient Descent with Different Learning Rates")
plt.xlabel("Iteration")
plt.ylabel("Value of x")
plt.grid(True)
plt.legend()

plt.show()
```

5. Write a short Python function that generates and plots a graph of f(x) = x*4 - 3x*3 + 2 for x in range(-2, 4), and visually identify one local minimum and the global minimum on the plot.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**4 - 3*x**3 + 2
```

```python
x = np.linspace(-2, 4, 200)
y = f(x)
```

```python
local_min_x = 2.25
local_min_y = f(local_min_x)
```

```python
plt.figure(figsize=(8, 6))
plt.plot(x, y, label="f(x) = x^4 - 3x^3 + 2")

plt.scatter(local_min_x, local_min_y, color="red", label="Minimum")
plt.annotate(
    "Local & Global Minimum",
    (local_min_x, local_min_y),
    textcoords="offset points",
    xytext=(10, 10)
)

plt.title("Graph of f(x) = x^4 - 3x^3 + 2")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()

plt.show()
```
