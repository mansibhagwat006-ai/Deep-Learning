2. Given a simple neural network with one hidden layer (using Keras), print out the weights before and after one training step on dummy data to observe how the weight update rule works.

```python
model = Sequential([
    Dense(4, activation='relu', input_shape=(2,)),
    Dense(1,activation='sigmoid')
])
```

```python
model.compile(optimizer='sgd',loss='binary_crossentropy')
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=float)

y = np.array([0, 1, 1, 0], dtype=float)
```

```python
print("Weights Before Training:\n")
for i, layer in enumerate(model.layers):
    print(f"Layer {i+1}")
    weights, bias = layer.get_weights()
    print("Weights:\n", weights)
    print("Bias:\n", bias)
    print()
```

```python
model.fit(X,y, epochs = 1 , batch_size = 4, verbose = 0)
```

```python
print("\nWeights After One Training Step:\n")
for i, layer in enumerate(model.layers):
    print(f"Layer {i+1}")
    weights, bias = layer.get_weights()
    print("Weights:\n", weights)
    print("Bias:\n", bias)
    print()
```

Manually calculate the gradient of the loss with respect to the weights for a single neuron using the chain rule, given these values: input x=2, weight w=0.5, bias b=1, activation=sigmoid, target y=1.<br><br><em><strong>Hint:</strong> Write out each step of the derivative using the chain rule, then compute the numeric value.</em>

```python
import math
```

```python
x = 2
w = 0.5
b = 1
y = 1
```

```python
z = w * x + b
a = 1/(1+math.exp(-z))
```

```python
loss = 0.5 * (a-y)**2
```

```python
dL_da = a - y
da_dz = a * (1 - a)
dz_dw = x
gradient = dL_da * da_dz * dz_dw
```

```python
print("Weighted Sum (z):", z)
print("Sigmoid Output (a):", round(a, 4))
print("Loss:", round(loss, 4))
print("dL/da:", round(dL_da, 4))
print("da/dz:", round(da_dz, 4))
print("dz/dw:", dz_dw)
print("Gradient (dL/dw):", round(gradient, 4))
```

4. Build a function in Python that simulates vanishing gradient by applying the sigmoid activation repeatedly to an input (e.g., 0.9) for 10 layers, and print the final output.
<br><br><em><strong>Hint:</strong> Observe how the value shrinks as you increase the number of layers.</em>

```python
import math
```

```python
def sigmoid(x):
    return 1 / (1+math.exp(-x))
```

```python
def vanishing_gradient(input_value , layers):
    value = input_value

    print("Layer\tOutput")
    for i in range(1,layers+1):
        value = sigmoid(value)
        print(f"{i}\t{value:.6f}")    
    return value
```

```python
input_value = 0.9
```

```python
final_output = vanishing_gradient(input_value, 10)
```

```python
print("\nFinal Output after 10 layers:", final_output)
```

5. Use ChatGPT or Copilot to generate Python code that demonstrates exploding gradients in a deep network by initializing weights with large values, then explain in 2-3 lines what happens to the gradients during backpropagation.

```python
import numpy as np

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Large weight initialization
weights = [10, 10, 10, 10, 10]

# Input
x = 1.0

print("Layer\tOutput")

# Forward pass through 5 layers
for i, w in enumerate(weights, start=1):
    x = sigmoid(w * x)
    print(f"{i}\t{x:.6f}")

# Simulated gradient during backpropagation
gradient = 1.0

print("\nBackpropagation Gradients:")

for i, w in enumerate(reversed(weights), start=1):
    gradient = gradient * w
    print(f"Layer {len(weights)-i+1}: {gradient:.2f}")
```

Large weight values cause the gradients to grow rapidly during backpropagation. As the gradient is multiplied by these large weights across many layers, it can become extremely large (explode), leading to unstable training, very large weight updates, and possible divergence of the model.
