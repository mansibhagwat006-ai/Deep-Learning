1. Implement the Leaky ReLU activation function in Python as a function leaky_relu(x, alpha=0.01), and test it on an array of values ranging from -5 to 5.

```python
import numpy as np
```

```python
def leaky_relu(x,alpha=0.01):
    return np.where(x>0,x,alpha*x)
```

```python
x =np.arange(-5,6)
```

```python
output = leaky_relu(x)
```

```python
print("Input:",x)
print("Output:",output)
```

2. Write a Python function parametric_relu(x, a) where 'a' is a learnable parameter (try a=0.2), and plot its output for x values from -5 to 5 using matplotlib.<br><br><em><strong>Hint:</strong> Compare the graph with standard ReLU and Leaky ReLU on the same plot for better understanding.</em>

```python
import matplotlib.pyplot as plt
```

```python
def parametric_relu(x,a):
    return np.where(x>0,x,a*x)

def relu(x):
    return np.maximum(0,x)

def leaky_relu(x,alpha=0.01):
    return np.where(x>0,x,alpha*x)
```

```python
x = np.linspace(-5,5,100)
```

```python
y_prelu = parametric_relu(x,0.5)
y_relu = relu(x)
y_lrelu = leaky_relu(x,0.1)
```

```python
plt.figure(figsize=(8,6))

plt.plot(x,y_prelu,label= "Parametric Relu(alpha = 0.5)")
plt.plot(x,y_relu,label ="Relu")
plt.plot(x,y_lrelu,label ="Leaky Relu(alpha = 0.1)")

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.xlabel("Input")
plt.ylabel("Activation Output")
plt.title("Relu VS Leaky Relu VS Parametric Relu")

plt.legend()
plt.grid(True)
plt.show()
```

3. Given a list of predicted scores for 4 classes: [2.0, 1.0, 0.1, 0.5], implement the softmax function in Python and print the resulting probabilities rounded to 3 decimal places.

```python
score = np.array([2.0, 1.0, 0.1, 0.5])
```

```python
def softmax(x):
    exp_x = np.exp(x)
    return exp_x/np.sum(exp_x)
```

```python
probabilities = softmax(score)
```

```python
print(np.round(probabilities,3))
```

4. Write a function hard_sigmoid(x) that implements the hard sigmoid activation (output between 0 and 1), and test it on inputs: -3, -1, 0, 1, 3.<br><br><em><strong>Hint:</strong> Use the formula: hard_sigmoid(x) = max(0, min(1, 0.2*x + 0.5)).</em>

```python
def hard_sigmoid(x):
    return max(0, min(1, 0.2*x + 0.5))
```

```python
inputs = [-3,-1,0,1,3]
```

```python
for x in inputs:
    print(f"Inputs: {x}, Hard sigmoid output :{hard_sigmoid(x)}")
```

5. Suppose you are building a multi-class image classifier for food items (like Zomato's food image feed). For each layer below, choose and justify the best activation function: (a) hidden dense layer, (b) output layer with 5 classes, (c) output layer with binary classification.<br><br><em><strong>Constraint:</strong> Write 1-2 lines of reasoning for each choice.</em>

#### Activation Functions for a Multi-Class Image Classifier

#### (a) Hidden Dense Layer
**Activation Function:** ReLU (Rectified Linear Unit)

**Reason:**  
ReLU is computationally efficient and helps reduce the vanishing gradient problem. It enables the model to learn complex features quickly.

#### (b) Output Layer with 5 Classes
**Activation Function:** Softmax

**Reason:**  
Softmax converts the output scores into probabilities that sum to 1, making it ideal for multi-class classification where only one class is predicted.

#### (c) Output Layer with Binary Classification
**Activation Function:** Sigmoid

**Reason:**  
Sigmoid outputs a probability between 0 and 1, making it suitable for binary classification tasks with two possible classes.

#### Summary

| Layer | Best Activation Function | Reason |
|--------|--------------------------|--------|
| Hidden Dense Layer | **ReLU** | Fast, efficient, reduces vanishing gradient, and learns complex features effectively. |
| Output Layer (5 Classes) | **Softmax** | Produces probabilities that sum to 1, making it ideal for multi-class classification. |
| Output Layer (Binary Classification) | **Sigmoid** | Produces a probability between 0 and 1, making it suitable for binary classification. |
