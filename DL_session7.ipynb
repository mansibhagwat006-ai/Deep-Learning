1. Write a Python function calculate_mse(predictions, targets) that takes two lists of numbers and returns the Mean Squared Error between them using the MSE formula.

```python
def calculate_mse(predictions, target):
    n= len(predictions)
    mse = 0 

    for i in range(n):
        mse += (predictions[i]- target[i])**2

    mse = mse/n
    return mse
```

```python
predictions = [3, 0.5,2,1]
target = [2,1,4,1]
```

```python
result = calculate_mse(predictions , target)
```

```python
print("Mean Squared Error:",result)
```

2. Given two arrays: actual_ratings = [4, 3, 5, 2, 1] and predicted_ratings = [5, 2, 4, 2, 1], calculate the Mean Absolute Error (MAE) using Python and print the result.

```python
actual_ratings = [4, 3, 5, 2, 1] 
predicted_ratings = [5, 2, 4, 2, 1]
```

```python
def calculate_mae(predictions , target):
    n = len(predictions)
    mae = 0

    for i in range(n):
        mae += abs(predictions[i]-target[i])

    mae = mae/n
    return mae 
```

```python
result = calculate_mae(predicted_ratings,actual_ratings)
```

```python
print("Mean Absolute Error:",result)
```

3. Implement a function binary_cross_entropy(y_true, y_pred) in Python that computes the binary cross-entropy loss for a batch of predictions and true labels (both as lists of 0s and 1s), then test it with y_true = [1, 0, 1] and y_pred = [0.9, 0.2, 0.7].

```python
import math 
```

```python
def binary_cross_entropy(y_true, y_pred):
    n = len(y_true)
    loss = 0

    for i in range(n):
        loss += -(y_true[i]*math.log(y_pred[i]) + (1-y_true[i]) * math.log(1- y_pred[i]))

    loss = loss/n
    return loss
```

```python
y_true =[1,0,1]
y_pred =[0.9,0.2,0.7]
```

```python
Error = binary_cross_entropy(y_true,y_pred)
```

```python
print("The loss :", Error)
```

4. Simulate a basic Categorical Cross-Entropy calculation: given y_true = [0, 0, 1] and y_pred = [0.2, 0.3, 0.5], write Python code to compute the categorical cross-entropy loss for this single data point.

```python
def cat_cross_entropy(y_true,y_pred):
    n = len(y_true)
    loss = 0

    for i in range(n):
        loss += y_true[i]* math.log(y_pred[i])

    loss = -loss
    return loss
```

```python
y_true = [0, 0, 1] 
y_pred = [0.2, 0.3, 0.5]
```

```python
error = cat_cross_entropy(y_true, y_pred)
```

```python
print("Categerical Cross entropy:",error)
```

5. Pick any app you use daily (like Zomato, Spotify, or Instagram) and describe a real-world scenario where choosing the right loss function (MSE, MAE, Binary Cross-Entropy, or Categorical Cross-Entropy) would be important. Explain which loss function you would use and why

App: Spotify

Scenario:
Spotify recommends songs based on whether a user is likely to like or skip a song. The model predicts the probability that the user will like a recommended song.

Loss Function: Binary Cross-Entropy (BCE)

Why?

The outcome has two classes: Like (1) or Not Like (0).
Binary Cross-Entropy measures how well the predicted probabilities match the actual user actions.
It penalizes incorrect predictions more strongly when the model is very confident but wrong, helping the model learn more effectively.

Example:

Actual: User likes the song → 1
Predicted probability: 0.95 → Low loss (good prediction)
Predicted probability: 0.10 → High loss (poor prediction)

Conclusion:
Binary Cross-Entropy is the best choice because Spotify's recommendation task is a binary classification problem where the model predicts whether a user will like a song or not.
