4. Compile your model using the 'adam' optimizer, 'categorical_crossentropy' loss, and track 'accuracy' as a metric. Then, train the model for 5 epochs using random data shaped like CIFAR-10 images and labels.<br><br><em><strong>Constraint:</strong> Use numpy to generate random input data and one-hot encoded labels for testing if you don't want to download the full CIFAR-10 dataset.</em>

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    BatchNormalization,
    MaxPooling2D,
    Flatten,
    Dropout,
    Dense
)
from tensorflow.keras.utils import to_categorical
```

```python
model = Sequential()

model.add(Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2, 2)))
model.add(Flatten())
model.add(Dropout(0.3))
model.add(Dense(64, activation='relu'))
model.add(Dense(10, activation='softmax'))
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
X_train = np.random.rand(100, 32, 32, 3).astype(np.float32)
y_train = np.random.randint(0, 10, 100)
y_train = to_categorical(y_train, num_classes=10)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=16,
    verbose=1
)
```

```python
print("\nTraining Accuracy:", history.history['accuracy'])
print("Training Loss:", history.history['loss'])
```

5. Use ChatGPT or Copilot to suggest one improvement to your CNN architecture for better accuracy on image classification, and implement the suggested change in your code.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    BatchNormalization,
    MaxPooling2D,
    Flatten,
    Dropout,
    Dense
)
import numpy as np
from tensorflow.keras.utils import to_categorical
```

```python
model = Sequential()

model.add(Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Conv2D(64, (3, 3), activation='relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2, 2)))
```

```python
model.add(Flatten())
model.add(Dropout(0.3))
model.add(Dense(64, activation='relu'))
model.add(Dense(10, activation='softmax'))
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
X_train = np.random.rand(100, 32, 32, 3).astype(np.float32)
y_train = np.random.randint(0, 10, 100)
y_train = to_categorical(y_train, num_classes=10)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=16,
    verbose=1
)
```

```python
model.summary()
```

```python
print("\nTraining Accuracy:", history.history["accuracy"])
print("Training Loss:", history.history["loss"])
```

### Suggested Improvement

**Improvement:** Added a second **Conv2D + BatchNormalization + MaxPooling2D** block.

**Reason:** The additional convolution layer enables the CNN to learn more complex image features (such as shapes and textures) after detecting simple edges. This generally improves feature extraction and can lead to better image classification accuracy on datasets like CIFAR-10.
