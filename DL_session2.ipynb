1. Create a Python script using NumPy to implement a perceptron that learns to classify whether a 
Flipkart product review is 'positive' (1) or 'negative' (0) based on two features: review length and 
number of exclamation marks. Use a small manually created dataset for training.

```python
import numpy as np
```

```python
x=np.array([
    [0.9,0.8],
    [0.6,0.4],
    [0.3,0.8],
    [0.1,0.9],
    [0.5,0.2],
    [0.8,0.8],
    [0.1,0.7]
])
```

```python
y=np.array([1, 1, 0,  0,  0,  1,  0])
```

```python
weights = np.zeros(x.shape[1])
bias = 0
learning_rate = 0.1
epochs = 20
```

```python
def activation(a):
    if a>=0:
        return 1
    else :
        return 0
```

```python
for epoch in range(epochs):
    for i in range(len(x)):
        linear_output = np.dot(x[i],weights)+bias
        prediction = activation(linear_output)
        error = y[i]-prediction

        weights += learning_rate * error * x[i]
        bias += learning_rate * error 
```

```python
print("Final Weights :" , weights)
print("Final Bias :", bias)
```

```python
test_reviews = np.array([
    [0.85, 0.7],   
    [0.25, 0.1],   
    [0.75, 0.9],   
    [0.30, 0.20]     
])
```

```python
print("\nPredictions :")
for review in test_reviews:
    output = np.dot(review,weights) + bias
    prediction = activation(output)

    if prediction == 1:
        print(review , ": Positive review")
    else:
        print(review , ": Negetive review")
```

2. Write a function in Python that performs the weighted sum and applies a step activation function
 for a perceptron. Test your function with sample weights, inputs, and bias values, and print the 
output for each case.<br><br><em><strong>Hint:</strong> Use NumPy arrays for vectorized 
operations.</em>

```python
def step_activation(a):
    if a>=0:
        return 1
    else:
        return 0 
```

```python
def perceptron(input, weights, bias):
    weighted_sum = np.dot(input,weights) + bias
    output = step_activation(weighted_sum)
    return weighted_sum , output
```

```python
test_cases = [
    (np.array([1,1]),np.array([0.5 , 0.5]),-0.5),
    (np.array([0,1]),np.array([0.5 , 0.5]),-0.5),
    (np.array([1,0]),np.array([0.5 , 0.5]),-0.5),
    (np.array([0,0]),np.array([0.5 , 0.5]),-0.5)
]
```

```python
for i, (inputs , weights ,bias) in enumerate (test_cases, start= 1):
    weighted_sum , output = perceptron(inputs , weights , bias)
    print(f"Test case:{i}")
    print("Inputs:",inputs)
    print("Weights:",weights)
    print("Bias:",bias)
    print("Weighted sum:",weighted_sum)
    print("Output:",output)
```

3. Demonstrate the limitation of a single-layer perceptron by attempting to train it (using your code) to solve the XOR problem (inputs: [0,0], [0,1], [1,0], [1,1]; outputs: 0, 1, 1, 0). Show and explain why it fails to classify correctly.

```python
import numpy as np
```

```python
X=np.array([[0,0],[0,1],[1,0],[1,1]])
```

```python
y=np.array([0,1,1,0])
```

```python
weights=np.zeros(2)
bias=0
learning_rate=0.1
epochs =20
```

```python
def step(x):
    if x>=0:
        return 1
    else:
        return 0
```

```python
for epoch in range(epochs):
    for i in range(len(X)):
        weighted_sum = np.dot(X[i],weights)+bias
        prediction= step(weighted_sum)
        error=y[i]-prediction
        weights += learning_rate*error*X[i]
        bias += learning_rate*error
```

```python
print("Final weights:",weights)
print("Final bias:",bias)
```

```python
print("Predictions:")
for i in range(len(X)):
    weighted_sum = np.dot(X[i],weights)+bias
    prediction= step(weighted_sum)
    print(f"Input : {X[i]} Actual : {y[i]} Predicted : {prediction}")
```

Conclusion:   
-->A single-layer perceptron fails to solve the XOR problem because XOR is not linearly separable.  
-->Solving XOR requires a Multi-Layer Perceptron (MLP) with at least one hidden layer and a non-linear activation function.  

4. Implement a simple Multilayer Perceptron (MLP) in Python using only NumPy that can solve the XOR problem. Your network should have an input layer, one hidden layer, and an output layer, with manual forward propagation (no external libraries<br><br><em><strong>Hint:</strong> Use sigmoid activation in the hidden and output layers.</em>

```python
import numpy as np
```

```python
X=np.array([[0,0],[0,1],[1,0],[1,1]])
```

```python
y=np.array([[0],[1],[1],[0]])
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
def sigmoid_derivative(x):
    return x*(1-x)
```

```python
np.random.seed(42)
```

```python
input_size=2
hidden_size=2
output_size=1
```

```python
weights_input_hidden = np.random.rand(input_size,hidden_size)
bias_hidden = np.random.rand(1,hidden_size)

weights_hidden_output = np.random.rand( hidden_size, output_size)
bias_output = np.random.rand(1,output_size)
```

```python
learning_rate = 0.1
epochs = 10000
y
```

```python
for epoch in range(epochs):
    hidden_input=np.dot(X, weights_input_hidden) + bias_hidden
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output , weights_hidden_output) + bias_output
    predicted_output = sigmoid(final_input)

    error = y- predicted_output

    d_output = error * sigmoid_derivative(predicted_output)
    hidden_error = np.dot(d_output, weights_hidden_output.T)
    d_hidden = hidden_error * sigmoid_derivative(hidden_output)


    weights_hidden_output += np.dot(hidden_output.T , d_output) * learning_rate
    bias_output += np.sum(d_output, axis=0, keepdims=True) * learning_rate

    weights_input_hidden += np.dot(X.T , hidden_error * sigmoid_derivative(hidden_output)) * learning_rate
    bias_hidden += np.sum(d_hidden , axis=0 ,keepdims = True)* learning_rate
```

```python
print("Predicted Outputs:")
for i in range(len(X)):
    hidden = sigmoid(np.dot(X[i], weights_input_hidden) + bias_hidden)
    output = sigmoid(np.dot(hidden, weights_hidden_output) + bias_output)

    print(f"Input: {X[i]}  Output: {output[0][0]:.4f}  Predicted Class: {int(output[0][0] >= 0.5)}")
```
