1. Create a simple Keras Sequential model for classifying handwritten digits (use the MNIST dataset), and train it without any regularization. Save the validation accuracy after 5 epochs.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
(X_train , y_train) , (X_test , y_test) = tf.keras.datasets.mnist.load_data()
```

```python
X_train = X_train/255.0
x_test = X_test/255.
```

```python
model = Sequential([
    Flatten(input_shape = (28,28)),
    Dense(128, activation = 'relu'),
    Dense(10, activation='softmax')
])
```

```python
model.compile(optimizer = 'adam', loss = 'sparse_categorical_crossentropy',
              metrics = ['accuracy'])
```

```python
history = model.fit(X_train , y_train , epochs = 5, validation_data = (X_test,y_test))
```

```python
val_accuracy = history.history['val_accuracy']
```

```python
print("Validation Accuracy after 5 epochs:",val_accuracy)
```

2. Modify your Keras model to add L2 regularization (kernel_regularizer=regularizers.l2(0.01)) to all Dense layers, retrain for 5 epochs, and compare the validation accuracy to your previous result

```python
from tensorflow.keras import regularizers
```

```python
model = Sequential([
    Flatten(input_shape = (28,28)),
    Dense(128,activation="relu",kernel_regularizer= regularizers.l2(0.01)),
    Dense(10, activation='softmax', kernel_regularizer = regularizers.l2(0.01))
])
```

```python
model.compile(optimizer='adam', loss = 'sparse_categorical_crossentropy',
              metrics =['accuracy'])
```

```python
history = model.fit(X_train, y_train, epochs = 5, validation_data = (x_test, y_test))
```

```python
val_accuracy = history.history['val_accuracy']
```

```python
print("Validation Accuracy after 5 epochs:",val_accuracy)
```

| Model | Validation Accuracy (Epoch 1 → 5) | Observation |
|-------|------------------------------------|-------------|
| **Without Regularization** | **[0.9570, 0.9715, 0.9710, 0.9742, 0.9770]** | Validation accuracy steadily improves over the 5 epochs, reaching **97.70%**. The model learns quickly but may be more prone to overfitting. |
| **With L2 Regularization (`regularizers.l2(0.01)`)** | **[0.9108, 0.9092, 0.9071, 0.9101, 0.9074]** | Validation accuracy remains around **91%** throughout training. The strong L2 penalty limits the model's complexity, resulting in lower accuracy but reducing the risk of overfitting. |

3. Add Dropout layers (with rate=0.5) after each Dense layer in your Keras model, retrain for 5 epochs, and observe the change in validation accuracy.<br><br><em><strong>Hint:</strong> Use keras.layers.Dropout and place it after activations.</em>

```python
from tensorflow.keras.layers import Dropout
```

```python
model = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation = 'relu'),
    Dropout(0.5),
    Dense(10,activation = 'softmax'),
    Dropout(0.5)
])
```

```python
model.compile(optimizer = 'adam', loss = 'sparse_categorical_crossentropy', metrics =['accuracy'])
```

```python
history = model.fit(X_train , y_train , epochs = 5 , validation_data= (X_test, y_test))
```

```python
val_accuracy = history.history['val_accuracy']
```

```python
print('Validation Accuracy after 5 epochs and adding dropout:', val_accuracy)
```

4. Integrate BatchNormalization layers into your Keras model (before activation functions in Dense layers), retrain, and note how the training and validation accuracy curves change.<br><br><em><strong>Hint:</strong> Use keras.layers.BatchNormalization before the activation in each Dense block.</em>

```python
from tensorflow.keras.layers import BatchNormalization, Activation
```

```python
model = Sequential([
    Flatten(input_shape=(28,28)),

    Dense(128),
    BatchNormalization(),
    Activation('relu'),

    Dense(10),
    BatchNormalization(),
    Activation('softmax')
])
```

```python
model.compile(optimizer = 'adam', loss ='sparse_categorical_crossentropy', metrics =['accuracy'])
```

```python
history = model.fit(X_train , y_train , epochs = 5 , validation_data = (X_test, y_test))
```

```python
print('Training Accuracy:', history.history['accuracy'])
print('Validation Accuracy:', history.history['val_accuracy'])
```

| Model | Training Accuracy (Epoch 1 → 5) | Validation Accuracy (Epoch 1 → 5) | Observation |
|-------|----------------------------------|------------------------------------|-------------|
| With Batch Normalization | **[0.9093, 0.9514, 0.9637, 0.9691, 0.9745]** | **[0.8411, 0.8080, 0.8374, 0.8184, 0.8112]** | The training accuracy increased steadily, but the validation accuracy remained much lower and even decreased over time. This indicates that the model is not generalizing well to the validation data and may be overfitting or using an unsuitable Batch Normalization configuration. |

5. Implement EarlyStopping in your model training using keras.callbacks.EarlyStopping with patience=2, and show how it affects the number of training epochs and final validation accuracy.

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128,activation='relu'),
    Dense(10,activation='softmax')
])
```

```python
model.compile(optimizer = 'adam', loss = 'sparse_categorical_crossentropy', metrics =['accuracy'])
```

```python
early_stopping = EarlyStopping(monitor= 'val_loss', patience = 2,restore_best_weights = True)
```

```python
history = model.fit(
    X_train , y_train ,
    epochs = 5 ,
    validation_data =(X_test, y_test),
    callbacks=[early_stopping]
    )
```

```python
print("Number of epochs trained:", len(history.history['loss']))
print("Final Validation Accuracy:", history.history['val_accuracy'][-1])
```

| Without Early Stopping | With Early Stopping (patience=2) |
|-------------------------|----------------------------------|
| Trains for all specified epochs (e.g., 20), even if validation performance stops improving. | Stops training automatically when the validation loss does not improve for 2 consecutive epochs. |
| May lead to overfitting and unnecessary training time. | Reduces overfitting, saves training time, and restores the best model weights. |
| Final validation accuracy depends on the last epoch. | Final validation accuracy corresponds to the best-performing model before training stopped. |
