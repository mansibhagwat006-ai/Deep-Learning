2. Convert a colored image (choose any photo you like) to grayscale and then back to RGB using Python libraries. Display all three versions (original, grayscale, RGB) using matplotlib.

```python
from PIL import Image
import matplotlib.pyplot as plt
```

```python
image = Image.open(r"E:\TOPS-DS\virat_kohli.jpeg")
original = image.convert("RGB")
grayscale = original.convert("L")
gray_to_rgb = grayscale.convert("RGB")
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(original)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(grayscale, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(gray_to_rgb)
plt.title("Grayscale to RGB")
plt.axis("off")

plt.tight_layout()
plt.show()
```

4. Write code to increase and decrease the brightness of a selfie or any image by 40% using ImageDataGenerator's brightness_range parameter. Show the original, brightened, and darkened images side by side.

```python
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
```

```python
image_path = r"E:\TOPS-DS\virat_kohli.jpeg"   
image = load_img(image_path)
image_array = img_to_array(image)
image_array = image_array.reshape((1,) + image_array.shape)
```

```python
bright_gen = ImageDataGenerator(brightness_range=[1.4, 1.4]) 
dark_gen = ImageDataGenerator(brightness_range=[0.6, 0.6])    
```

```python
bright_image = next(bright_gen.flow(image_array, batch_size=1))[0].astype("uint8")
dark_image = next(dark_gen.flow(image_array, batch_size=1))[0].astype("uint8")
```

```python
original_image = image_array[0].astype("uint8")
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(original_image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(bright_image)
plt.title("Brightened (+40%)")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(dark_image)
plt.title("Darkened (-40%)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

5. Pick any meme or cricket image and use ImageDataGenerator to apply a shear transformation. Explain in one line how the image changes visually after shearing.

```python
image_path = r"E:\TOPS-DS\virat_kohli.jpeg"   
image = load_img(image_path)
image_array = img_to_array(image)
image_array = image_array.reshape((1,) + image_array.shape)
```

```python
datagen = ImageDataGenerator(shear_range=25)
sheared_image = next(datagen.flow(image_array, batch_size=1))[0].astype("uint8")
original_image = image_array[0].astype("uint8")
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(original_image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(sheared_image)
plt.title("Sheared Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
