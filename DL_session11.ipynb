1. Train a simple image classifier on the Fashion MNIST dataset using Keras and the SGD optimizer with default parameters, then record the accuracy and loss after 5 epochs.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten , Dense , Activation
from tensorflow.keras.optimizers import SGD
```

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
```

```python
X_train = X_train/255.0
X_test = X_test/255.0
```

```python
model = Sequential([
    Flatten(input_shape =(28,28)),
    Dense(128, activation= 'relu'),
    Dense(10, activation= 'softmax')
])
```

```python
model.compile(optimizer ='SGD', loss = 'sparse_categorical_crossentropy', metrics =['accuracy'])
```

```python
history = model.fit(X_train , y_train , epochs = 5 , validation_data = (X_test, y_test))
```

```python
print("Accuracy:",history.history['accuracy'])
print("Loss:",history.history['loss'])
print("Validation Accuracy:",history.history['val_accuracy'])
print("Validation loss:",history.history['val_loss'])
```

4. Implement a learning rate schedule in your Keras model using the LearningRateScheduler callback to reduce the learning rate by half after every 2 epochs, and train with the Momentum optimizer; report how the learning rate changes during training.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.optimizers import SGD
from tensorflow.keras.callbacks import LearningRateScheduler
```

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
```

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
optimizer = SGD(learning_rate=0.01, momentum=0.9)
```

```python
model.compile(
    optimizer=optimizer,
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
def schedule(epoch, lr):
    if epoch > 0 and epoch % 2 == 0:
        return lr * 0.5
    return lr
```

```python
lr_scheduler = LearningRateScheduler(schedule, verbose = 1)
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=6,
    validation_data=(X_test, y_test),
    callbacks=[lr_scheduler]
)
```

```python
learning_rates = []
lr = 0.01
```

```python
for epoch in range(6):
    if epoch > 0 and epoch % 2 == 0:
        lr *= 0.5
    learning_rates.append(lr)
```

```python
print("\nLearning Rate at Each Epoch:")
for epoch, lr in enumerate(learning_rates, start=1):
    print(f"Epoch {epoch}: {lr}")
```

5. Use ChatGPT to explain, in your own words, why Adam is often preferred in industry projects over SGD or RMSProp, and give one real-world app scenario (like Zomato, Flipkart, or Spotify) where optimizer choice could impact user experience.

##### Why Adam is Often Preferred

Adam (Adaptive Moment Estimation) is widely used in industry because it combines the advantages of Momentum and adaptive learning rates. It converges faster, requires less manual tuning of the learning rate, and performs well on large and complex datasets. Compared to SGD and RMSProp, Adam generally provides more stable and efficient training, making it a popular choice for deep learning applications.

##### Real-World Scenario

**App:** Spotify

Spotify uses deep learning to recommend songs and playlists based on a user's listening history. Choosing an efficient optimizer like **Adam** helps the recommendation model train faster and achieve better accuracy. As a result, users receive more relevant music recommendations, improving their overall listening experience and satisfaction.
