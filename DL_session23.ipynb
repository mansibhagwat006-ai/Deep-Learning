1. Load the ResNet50 model with pretrained ImageNet weights using Keras, print the model summary, and count how many layers are trainable by default.

```python
from tensorflow.keras.applications import ResNet50
```

```python
model = ResNet50(
    weights='imagenet',
    include_top=True
)
```

```python
model.summary()
```

```python
trainable_layers = sum(layer.trainable for layer in model.layers)
```

```python
print("\nTotal Layers:", len(model.layers))
print("Trainable Layers:", trainable_layers)
print("Non-Trainable Layers:", len(model.layers) - trainable_layers)
```

2. Use the VGG16 model (with include_top=False) as a feature extractor to process a batch of 10 images from any public dataset (e.g., CIFAR-10 or your own downloaded images), and print the shape of the extracted features.<br><br><em><strong>Hint:</strong> Use keras.applications.vgg16.preprocess_input() before passing images to the model.</em>

```python
import numpy as np
from tensorflow.keras.applications import VGG16
from tensorflow.keras.applications.vgg16 import preprocess_input
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.layers import Resizing
```

```python
(x_train, y_train), (_, _) = cifar10.load_data()
```

```python
images = x_train[:10]
resizer = Resizing(224, 224)
images_resized = resizer(images)
images_preprocessed = preprocess_input(images_resized)
```

```python
model = VGG16(
    weights='imagenet',
    include_top=False
)
```

```python
features = model.predict(images_preprocessed)
```

```python
print("Input Image Batch Shape:", images.shape)
print("Resized Image Batch Shape:", images_resized.shape)
print("Extracted Feature Shape:", features.shape)
```

3. Freeze all layers of the InceptionV3 model except the last 30 layers, then add a custom Dense output layer for 5-class classification, and print the number of trainable parameters.<br><br><em><strong>Hint:</strong> Use model.trainable = False, then set trainable=True for the last layers as needed.</em>

```python
import tensorflow as tf
from tensorflow.keras.applications import InceptionV3
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense
from tensorflow.keras.models import Model
```

```python
base_model = InceptionV3(
    weights='imagenet',
    include_top=False,
    input_shape=(299, 299, 3)
)
```

```python
base_model.trainable = False
```

```python
for layer in base_model.layers[-30:]:
    layer.trainable = True
```

```python
x = base_model.output
x = GlobalAveragePooling2D()(x)
output = Dense(5, activation='softmax')(x)
```

```python
model = Model(inputs=base_model.input, outputs=output)
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
trainable_params = sum(
    tf.keras.backend.count_params(w) for w in model.trainable_weights
)
```

```python
print("\nTotal Layers:", len(base_model.layers))
print("Trainable Layers:", sum(layer.trainable for layer in base_model.layers))
print("Trainable Parameters:", trainable_params)
```

5. Use ChatGPT or Copilot to generate Keras code that loads the InceptionV3 model, freezes all layers, adds a custom classification head with a Dropout layer, and compiles the model for binary classification. Paste the generated code and note any changes you made to get it running.

```python
import tensorflow as tf
from tensorflow.keras.applications import InceptionV3
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout
from tensorflow.keras.models import Model

# Load pretrained InceptionV3 model
base_model = InceptionV3(
    weights='imagenet',
    include_top=False,
    input_shape=(299, 299, 3)
)

# Freeze all layers
base_model.trainable = False

# Add custom classification head
x = base_model.output
x = GlobalAveragePooling2D()(x)

# Dropout to reduce overfitting
x = Dropout(0.5)(x)

# Binary classification output
output = Dense(1, activation='sigmoid')(x)

# Create final model
model = Model(
    inputs=base_model.input,
    outputs=output
)

# Compile model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Display model
model.summary()

# Count trainable parameters
trainable_params = sum(
    tf.keras.backend.count_params(w)
    for w in model.trainable_weights
)

print("Trainable Parameters:", trainable_params)
```

### Changes Made

- Used `include_top=False` to remove InceptionV3's original ImageNet classification layer.
- Set `base_model.trainable = False` to freeze all pretrained InceptionV3 layers.
- Added `GlobalAveragePooling2D()` to convert the extracted feature maps into a feature vector.
- Added `Dropout(0.5)` to help reduce overfitting.
- Added `Dense(1, activation='sigmoid')` because the task is binary classification.
- Used `binary_crossentropy` as the loss function, which is suitable for binary classification.
